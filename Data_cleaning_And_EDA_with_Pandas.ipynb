## Data Understanding

We are going to work with several columns selected from the main dataset:

**Numerical Columns:**
- **LotArea:** Total lot size in square feet
- **GrLivArea:** Above-ground living area in square feet
- **LotFrontage:** Linear feet of street connected to property
- **TotalBsmtSF:** Total basement area in square feet (0 if no basement)
- **GarageArea:** Size of garage in square feet (0 if no garage)
- **SalePrice:** Final sale price of the house

**Categorical Columns:**
- **MSZoning:** General zoning classification of the property  
  *Values:* A (Agriculture), C (Commercial), FV (Floating Village Residential), I (Industrial), RH (Residential High Density), RL (Residential Low Density), RP (Residential Low Density Park), RM (Residential Medium Density)
- **Neighborhood:** Physical location within Ames city limits  
- **Exterior1st:** Primary exterior covering of the house  
- **GarageType:** Garage location/type  
- **KitchenQual:** Kitchen quality rating  
  *Values:* Ex (Excellent), Gd (Good), TA (Typical/Average), Fa (Fair), Po (Poor)
- **ExterQual:** Exterior material quality rating  
- **BsmtQual:** Basement height/quality rating  
- **HeatingQC:** Heating quality and condition  
- **GarageQual:** Garage quality rating
